```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good'  
]
```

```python
## Define the vocabulary size
vocabulary_size = 10000
```

```python
## One Hot Representation
one_hot_repr = [one_hot(sentence, vocabulary_size) for sentence in sentences]
one_hot_repr
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sentence_length = 8
## all the sentences in the dataset must be of 8 words
```

```python
padded_sentences = pad_sequences(one_hot_repr, padding='pre', maxlen = sentence_length)
padded_sentences
```

```python
dimension = 10
```

```python
model = Sequential()
model.add(Embedding(vocabulary_size, dimension, input_length = sentence_length))

model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
model.predict(padded_sentences)
```
