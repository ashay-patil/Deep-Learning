```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
import pickle
```

```python
data = pd.read_csv("../dataset/Churn_Modelling.csv")
data.head()
```

```python
## Drop irrelavant columns
data = data.drop(['RowNumber', 'CustomerId', 'Surname'], axis = 1)
data.head()
```

```python
## Encoding Categorical Variables

lbl_encoder = LabelEncoder()
data['Gender'] = lbl_encoder.fit_transform(data['Gender'])

encoder = OneHotEncoder()
encoded = encoder.fit_transform(data[['Geography']]).toarray()
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out())

df = pd.concat([data, encoded_df], axis=1)

df.head()

```

```python
df = df.drop('Geography', axis = 1)
```

```python
## saving the encoders

with open('../models/label_encoder_gender.pkl', 'wb') as file : 
    pickle.dump(lbl_encoder, file)

with open('../models/onehot_encoder_geo.pkl', 'wb') as file : 
    pickle.dump(encoder, file)


```
